```python
from pypdf import PdfReader
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter 
from langchain_community.document_loaders import TextLoader, DirectoryLoader
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from dotenv import load_dotenv

load_dotenv()
```

```python
from pathlib import Path

def load_file_with_fallback(file_path, encodings=None):
    """Try multiple encodings until one works."""
    if encodings is None:
        # Try in this order (most common first for Python files)
        encodings = ["utf-8", "utf-8-sig", "latin-1", "cp1252"]
    
    for encoding in encodings:
        try:
            loader = TextLoader(str(file_path), encoding=encoding)
            return loader.load()
        except (UnicodeDecodeError, RuntimeError):
            continue
    
    # If all fail, log and return empty
    print(f"✗ Failed to load {file_path.name} (tried {encodings})")
    return []

project_path = Path("C:/Users/ZENOID/Desktop/Home/Startups/ESTATE/Estate Web")
skip_dirs = {"env", "venv", ".venv", "__pycache__"}

python_files = [
    f for f in project_path.rglob("*.py")
    if not any(skip in f.parts for skip in skip_dirs)
]

documents = []
for file_path in python_files:
    docs = load_file_with_fallback(file_path)
    if docs:
        documents.extend(docs)
        print(f"✓ {file_path.name}")

print(f"\nLoaded {len(documents)} documents")
```

```python
pdf_reader= PdfReader("C:/Users/ZENOID/Desktop/Home/Books/Django Full coducmentation.pdf")
docs=[
    Document(
        page_content= page.extract_text() or "",
        metadata= {"source": "django-cods.pdf", "page": i},
    )
    for i, page in enumerate(pdf_reader.pages)
]
```

```python
django_docs=docs
django_codebase=documents
```

```python
text_splitter= RecursiveCharacterTextSplitter(
    chunk_size= 1000, chunk_overlap=200, add_start_index=True
)
```

```python
text_code_splitter= RecursiveCharacterTextSplitter(
    chunk_size= 1500, chunk_overlap=500, add_start_index=True
)
```

```python
pdf_split= text_code_splitter.split_documents(django_docs)
```

```python
#Split Django COde
django_split= text_splitter.split_documents(django_codebase)
```

```python
print(f"Number of django docs chunks: {len(pdf_split)}")
print(f"Number of django codebase chunks: {len(django_split)}")
```

```python
#Creating out embedding model
embeddings= HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

```

* Creating a collection to seperate the docs and django script

```python
# Django Docs
def save_django_docs(chunked_django_doc):
    """Adds Django Documentation to a seperate collection"""
    vector_db = Chroma(
        persist_directory="./django_chroma_db",
        embedding_function=embeddings,
        collection_name="django_docs"
    )
    vector_db.add_documents(chunked_django_doc)
    print(f"Added {len(chunked_django_doc)} django chunks to - django_docs collection")
    return vector_db
```

```python
#Django Codebase
def save_django_code(chunked_django_code):
    """Add Chunked python script to a seperate collecton"""

    vector_db = Chroma(
        persist_directory="./django_chroma_db",
        embedding_function=embeddings,
        collection_name="python_scripts"
    )
    vector_db.add_documents(chunked_django_code)
    print(f"Added {len(chunked_django_code)} python chunks to - python_scripts collection")
    return vector_db
```

```python
django_docs= save_django_code(django_split)
django_code= save_django_docs(pdf_split)
```
